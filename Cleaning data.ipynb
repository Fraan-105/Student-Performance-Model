```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("student_performance.csv")

print("Dataset berhasil dimuat")
print(df.head())
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print("Jumlah data duplikat:", df.duplicated().sum())
```

```python
print(df["grade"].value_counts())
```

```python
df["grade"].value_counts().plot(kind="bar")

plt.title("Distribusi Grade Siswa")
plt.xlabel("Grade")
plt.ylabel("Jumlah Siswa")
plt.show()
```

```python
plt.subplot(1, 2, 2)
sns.scatterplot(data=df.sample(1000, random_state=42), x='attendance_percentage', y='total_score', hue='grade', hue_order=['A', 'B', 'C', 'D', 'F'], alpha=0.7)
plt.title('Total Score vs Attendance (Sample 1.000 Siswa)')
plt.xlabel('Attendance (%)')
plt.ylabel('Total Score')

plt.tight_layout()
plt.show()
```
